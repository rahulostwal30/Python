```python
"Reverse list"
l  = [1,2,3,4]
i = 0
n = len(l)-1
print(l)
while i<n:
    temp = l[n]
    l[n] = l[i]
    l[i] = temp
    i+=1
    n-=1
    print(l)
```

```python
" find all elements greater then the average"
n = [1,2,3,4,10]
sum = 0
count = 0
for i in range(len(n)):
    sum += n[i]
    count += 1
avg = sum//count

i=0
while i<len(n):
    if n[i] < avg:
        n.pop(i)
    else:
        i+=1
n
```

```python
"find the longest consective squence"
n = [1,2,3,4,5,10,11,12,13]
i = 1
count = 1
ma = 0
while i<len(n):
    if n[i] == n[i-1]+1:
        count += 1
    elif n[i] != n[i-1]:
        count = 1
    if ma < count:
        ma = count
    i+=1
    
ma
```

```python
"flatten a 2d list into a 1d list"
n = [[1,2],[3,4],[5,6]]
i = 0
j = len(n)-1

while j>=0:
    n.extend(n[i])
    n.pop(i)
    j-=1
n
```

```python
"list is a palindrom"

n = [1,2,3,2,1]
i = 0
j = len(n)-1
z = True
while i<j:
    if n[i] != n[j]:
        z = False
    i += 1
    j -= 1
if z == True:
    print("Palindrom")
else:
    print("NOt a palindrom")
```

```python
"find all pair sum to target value"
n = [1,2,3,4,5]
i = 0
j = 1
target = 5
new = []

while i<len(n)-1:
    if n[i]+n[j] == target:
        new.append((n[i],n[j]))
    j+=1
    if j==len(n):
        i+=1
        j = i+1
new
```

```python
"find the missing number in a list"
n = [1,2,4,5]
mn = min(n)
mx = max(n)
while mn<=mx:
    if mn not in n:
        print(mn)
    mn+=1
```
