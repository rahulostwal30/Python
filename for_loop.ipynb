```python
""" number is prime or not """
n = 19
count = 0
for i in range (2,n):
    if n%i==0:
        count+=1
if count==0:
    print(f" {n} is prime number")
else:
    print(f"{n} is not prime")
```

```python
""" Armstrong number"""
n = 153
temp = n
count = 0
new = 0
for i in range(len(str(n))):
    temp //= 10
    count+=1

for i in range(len(str(n))):
    rem = n%10
    new = new+(rem**count)
    n = n//10
new
```

```python
""" wheather a number is strong or not"""
n = 145
temp = n
sum = 0
for i in range(len(str(n))):
    rem = n%10
    out = 1
    for j in range(1,rem+1):
        out *= j
    sum += out
    n = n//10
if sum == temp :
    print("this number is strong")
else:
    print("this number is not strong")
```

```python
""" whether a number is perfect number or not"""
n = 6
temp = n
sm = 0
for i in range(1,n):
    if n%i==0:
        sm += i
if sm == temp :
    print("perfect")
```
