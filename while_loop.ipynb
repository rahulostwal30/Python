# While loop

A while loop is a control flow statement in python that repeatedly executes a block of code as long as a given condition is True. Think or it like a security guard who keeps checkinf a condition again and again until its no longer true 

. Key idea
1. A while loop says ; keep doinf this task while this condition is true
2. the moment the condition becomes false the loop stops
3. if the condition if false from the very begining the loop body never runs


1. The loop always returns to check the condition after each iteration.     
 if condition is true --> run body again ,   
 if false --> exit

# The 3 Essential parts



INITIALIZATION => set your counter/variable before the loop starts => i = 0

CONDITION => the test that decides whether to keep looping  => while i<10

UPDATE => change the variable so the condition eventually becomes false
           i+=1   without this use get infinite loop


# When to use while loop

1.  When you do NOT know in advance how many times to repeat
2.  When repeating until a specific event happens (e.g., user quits)
3.  When checking a condition dynamically at each step
4.  Game loops, waiting for user input, number processing

```python
"Q27. Remove Duplicate Characters"

word = "programming"
seen = ''
i = 0
while i < len(word):
    if word[i] not in seen:
        seen += word[i]
        
    i += 1
print(seen)
```
