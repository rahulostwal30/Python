#  ---------------       Inheritance           ---------------

Inheritance is a feature of oops 
where we used all the property and method of one class into another class

kind of inheritance :-
                    1. singal inheritance
                    2. multilevel inheritance
                    3. multipal inheritance

```python
class elderbrother:            # parent class/base class/super class
    salary = 10000

class ram(elderbrother):       # child class/ derived class / sub class
    pocket = 500



r1 = ram()
print(r1.pocket,r1.salary)     # child class is used to property of the parent class
```

```python
class elderbrother:
    salary = 10000

class ram(elderbrother):
    pocket = 5000

    def raminfo(self):
        print("Ram function : ",self.pocket,self.salary)


s1 = ram()
print(s1.raminfo())
```

```python
" Singal inheritance => A child class inherit from one parent class"

class parent:
    def display1(self):
        print("This is a parent class ")
    
class child(parent):
    def display2(self):
        print("This is a child class")

c1 = child()
c1.display2()
c1.display1()
```

```python
" Multilevel inheritance  => A child class can accessed the  method and property both the parent class and grandparent class "
"                            and parent class inherit grandparent class  "

class grandparent:
    def display1(self):
        print("This is a grandparent class")
class parent(grandparent):
    def display2(self):
        print("This is a parent class")
class child(parent):
    def display3(self):
        print("This is child class")

c2 = child()
c2.display3()
c2.display2()
c2.display1()

c3 = parent()
c3.display2()
c3.display1()
c3.display3()


```

```python
"Multiple inheritance => multiple parent class and one child class"
"                        a child class inherit from more then parent class"

class A:
    def display1(self):
        print("This is A class")
class B:
    def display2(self):
        print("This is B class")
class C(A,B):
    def display3(self):
        print("This is a one child class")

c4 = C()
c4.display3()
c4.display2()
c4.display1()
```

# Super() fuction

it is used to access methods and properties of a parents class from a child class

```python
class zomatoDriver:
    company = "Zomato pvt ltd"
    def __init__(self,name,email,ride):
        print("constructor runn")
        self.name = name
        self.email = email
        self.totalride = ride

class zomatocustomer(zomatoDriver):
    def __init__(self,x,y,z):
        print("----inside the constructor----")
        super().__init__(x,y,z)
        print("after calling the parent class constructor from the customer")



zd1 = zomatocustomer('raj','raj@gmail.com',23)
print("total ride :",zd1.totalride)
        
```

```python
"""
class variable can be accessed by both the class and object
        { class_name.company is a class variable }

instance variable can be accessed only by an object they cannot be accessed by class
        { self.name is a instance variable}
"""
```

```python
class driver:
    def __init__(self,name,email,trip):
        self.name = name
        self.email = email
        self.trip = trip

    def display(self):
        print("Driver class display : ",self.name,self.email)

class customer(driver):
    def __init__(self,name,email,trip):
        super().__init__(name,email,trip)

    def customer_display(self):
        print("information :",self.name,self.email)




alfa = customer('rajat','rajat@gmail',62)
print(alfa.name)
alfa.customer_display()
alfa.display()
```
