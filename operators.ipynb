#                              Operators 

```python
"""
a special symbal or keyword that perform mathamatically and non-mathamatically task
Type of operator ==> 1. Arithmetic operator [ +  -  *  %   /  //  ** ]
                     2. Camparision operator [ ==  !=  <  >  <=  >= ]
                     3. Assignment operator [-=  +=  *=  /=  %=  **=  =]
                     4. Logical operator  [ and  not  or ]
                     5. Bitwise operator [ <<  >>  &  |  ^  ~ ]
                     6. Membership operator [ in   not in]
                     7. Identity operator [ is  is not ]
"""
```

```python
"""
1. Arithmetic operator  => used to perform methematically task
priority => **
            / // % * (same priority left to right)
            + - 
""" 
print("Division :: ",20/6)
print("Addition ::",5+7)
print("substraction ::",8-3)
print("Mutliplication ::",2*3)
print("modulo ::",8%3)
print("floor division ::",10//2)
print("Exponent ::",2*3)

```

```python
"""
2. Camparision Operator / Relational operator 
"""
a = 5 ; b = 8
print(b>a)
print(b<a)
print(a<=b)
print(a>=b)
print(a!=b)
print(a==b)
```

```python
"""
3. Assignment operator 
"""
a = 10
a+=5
print(a)

a-=2
print(a)

a*=2
print(a)

a/=3
print(a)

a**=2
print(a)

a//=3
print(a)
```

```python

"""
4. Logical operator ==> used to multipal combine or expression ( and not or)
"""
a = 10

print("AND expression ::",a>50 and a>8)

print("OR expression ::",a!= 13 or a<8)


"""
5. Membership operator ==> used to check the value is present in the given set off data 
                           by using in or not in 
"""
print("u" in "Tushar")
print("x" not in "tushar")
```

```python

"""
6. Identity operator ==>  campare the memory address two or more objects
"""
x = 10
y = 10
print("Memory address of x => ",id(x))
print("memory address of y => ",id(y))
print(x is y)
print(x is not y)
```

```python

"""
7. Bitwise operator => operators on numbers 
                       integers
                       bits(0 1)
"""
print(7 & 13)

# Binary value of 7 => 0111
# binary value of 13 =>1101

print(7 | 13)
print("Left shift :",7<<2)
print("Right shift :",7>>2)
```
